```python
import pandas as pd
import numpy as np
import random

dataset = pd.read_csv("digital_addiction_dataset.csv")

print("DIGITAL ADDICTION ANALYZER")
print("Dataset loaded successfully!")
print("Total Users:", len(dataset))
print("Total Columns:", len(dataset.columns))
print("\nDataset Preview:")
print(dataset.head())
```

```python
screen = (dataset["Daily_Screen_Time"] / 12) * 30
social = (dataset["Social_Media_Hours"] / 6) * 15
gaming = (dataset["Gaming_Hours"] / 5) * 10
phone = (dataset["Phone_Checks"] / 150) * 10
notifications = (dataset["Notifications_Per_Day"] / 200) * 5

night = np.where(dataset["Night_Usage"] == "Yes", 10, 0)

sleep = np.where(
    dataset["Sleep_Hours"] < 6, 10,
    np.where(dataset["Sleep_Hours"] < 7, 5, 0)
)

productivity = np.where(
    dataset["Productivity_Score"] <= 4, 10,
    np.where(dataset["Productivity_Score"] <= 6, 5, 0)
)

dataset["Addiction_Score"] = (
    screen + social + gaming + phone +
    notifications + night + sleep + productivity
).round(2)

def addiction_level(score):
    if score < 35:
        return "Low"
    elif score < 65:
        return "Moderate"
    else:
        return "High"

dataset["Addiction_Level"] = dataset["Addiction_Score"].apply(addiction_level)

def screen_category(hours):
    if hours < 4:
        return "Low"
    elif hours < 7:
        return "Moderate"
    else:
        return "High"

dataset["Screen_Time_Category"] = dataset["Daily_Screen_Time"].apply(screen_category)

print("Addiction scores calculated successfully!")
print(dataset[["User_ID", "Addiction_Score", "Addiction_Level"]].head(10))
```

```python
print("=" * 50)
print("       DIGITAL ADDICTION ANALYSIS")
print("=" * 50)

print("\nDataset Statistics:")
print(dataset.describe().round(2))

print("\nAverage Screen Time:",
      round(dataset["Daily_Screen_Time"].mean(), 2), "hours")

print("Average Social Media:",
      round(dataset["Social_Media_Hours"].mean(), 2), "hours")

print("Average Gaming:",
      round(dataset["Gaming_Hours"].mean(), 2), "hours")

print("Average Sleep:",
      round(dataset["Sleep_Hours"].mean(), 2), "hours")

print("Average Productivity:",
      round(dataset["Productivity_Score"].mean(), 2))

print("\nAddiction Distribution:")
print(dataset["Addiction_Level"].value_counts())

print("\nAddiction Percentage:")
print((dataset["Addiction_Level"].value_counts(normalize=True) * 100).round(2))

print("\nNight Usage Analysis:")
print(dataset.groupby("Night_Usage")["Addiction_Score"].mean().round(2))

dataset["Age_Group"] = pd.cut(
    dataset["Age"],
    bins=[16, 18, 21, 24],
    labels=["17-18", "19-21", "22-24"]
)

print("\nAge Group Analysis:")
print(dataset.groupby("Age_Group", observed=True)["Addiction_Score"].mean().round(2))

print("\nGender Analysis:")
print(dataset.groupby("Gender")[
    ["Daily_Screen_Time", "Sleep_Hours", "Addiction_Score"]
].mean().round(2))

high_risk = dataset[
    (dataset["Daily_Screen_Time"] >= 7) &
    (dataset["Sleep_Hours"] < 6) &
    (dataset["Productivity_Score"] <= 4)
]

print("\nTotal High-Risk Users:", len(high_risk))

top_10 = dataset.nlargest(10, "Addiction_Score")

print("\nTOP 10 ADDICTED USERS:")
print(top_10[[
    "User_ID", "Daily_Screen_Time",
    "Sleep_Hours", "Addiction_Score", "Addiction_Level"
]])

arr = dataset["Daily_Screen_Time"].to_numpy()

print("\nNUMPY ANALYSIS")
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))

print("\nUnique Genders:", set(dataset["Gender"]))
print("Unique Addiction Levels:", set(dataset["Addiction_Level"]))
```

```python
def user_report(user_id):

    result = dataset[dataset["User_ID"].str.upper() == user_id.upper()]

    if result.empty:
        print("User not found!")
        return

    user = result.iloc[0]

    print("\n" + "=" * 45)
    print("       DIGITAL ADDICTION REPORT")
    print("=" * 45)

    print("User ID:", user["User_ID"])
    print("Age:", user["Age"])
    print("Gender:", user["Gender"])

    print("\nSCREEN TIME DETAILS")
    print("-" * 30)

    print("Daily Screen Time:", user["Daily_Screen_Time"], "hours")
    print("Social Media:", user["Social_Media_Hours"], "hours")
    print("Gaming:", user["Gaming_Hours"], "hours")
    print("Study:", user["Study_Hours"], "hours")

    print("\nLIFESTYLE DETAILS")
    print("-" * 30)

    print("Sleep:", user["Sleep_Hours"], "hours")
    print("Phone Checks:", user["Phone_Checks"])
    print("Notifications:", user["Notifications_Per_Day"])
    print("Night Usage:", user["Night_Usage"])
    print("Mood Score:", user["Mood_Score"])
    print("Productivity:", user["Productivity_Score"])

    print("\nADDICTION ANALYSIS")
    print("-" * 30)

    print("Addiction Score:", user["Addiction_Score"])
    print("Addiction Level:", user["Addiction_Level"])

    if user["Addiction_Level"] == "High":
        print("\nRecommendation: Reduce screen time and improve sleep habits.")

    elif user["Addiction_Level"] == "Moderate":
        print("\nRecommendation: Monitor daily digital usage.")

    else:
        print("\nRecommendation: Maintain healthy digital habits.")

    print("=" * 45)
```

```python
def main_menu():

    while True:

        print("\n========== DIGITAL ADDICTION ANALYZER ==========")
        print("1. Display Dataset")
        print("2. Overall Report")
        print("3. Addiction Distribution")
        print("4. Screen Time Analysis")
        print("5. Social Media and Gaming Analysis")
        print("6. Sleep and Productivity Analysis")
        print("7. High-Risk Users")
        print("8. Top 10 Addicted Users")
        print("9. Search User Report")
        print("10. Save All Reports")
        print("11. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print(dataset.head(20))

        elif choice == "2":
            print("\nTotal Users:", len(dataset))
            print("Average Screen Time:", round(dataset["Daily_Screen_Time"].mean(), 2))
            print("Average Addiction Score:", round(dataset["Addiction_Score"].mean(), 2))
            print("High-Risk Users:", len(high_risk))

        elif choice == "3":
            print(dataset["Addiction_Level"].value_counts())

        elif choice == "4":
            print(dataset["Daily_Screen_Time"].describe())

        elif choice == "5":
            print("Average Social Media:",
                  round(dataset["Social_Media_Hours"].mean(), 2))
            print("Average Gaming:",
                  round(dataset["Gaming_Hours"].mean(), 2))

        elif choice == "6":
            print("Average Sleep:",
                  round(dataset["Sleep_Hours"].mean(), 2))
            print("Average Productivity:",
                  round(dataset["Productivity_Score"].mean(), 2))

        elif choice == "7":
            print(high_risk[[
                "User_ID", "Daily_Screen_Time",
                "Sleep_Hours", "Productivity_Score",
                "Addiction_Level"
            ]])

        elif choice == "8":
            print(top_10[[
                "User_ID", "Addiction_Score", "Addiction_Level"
            ]])

        elif choice == "9":
            user_id = input("Enter User ID (e.g. U001): ")
            user_report(user_id)

        elif choice == "10":
            dataset.to_csv("digital_addiction_analyzed.csv", index=False)
            high_risk.to_csv("high_risk_users.csv", index=False)
            top_10.to_csv("top_10_addicted_users.csv", index=False)

            print("\nAll CSV reports saved successfully!")

        elif choice == "11":
            print("Thank you for using Digital Addiction Analyzer!")
            break

        else:
            print("Invalid choice! Please try again.")


main_menu()
```
